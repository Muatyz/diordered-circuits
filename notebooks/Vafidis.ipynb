- 数量
    - head-direction cells(头朝向细胞) 数量: $N_{\mathrm{HD}} = 60$, 形成 ring. 
    - head-rotation cells(头旋转细胞) 数量: $N_{\mathrm{HR}} = 60$
        - left HR cells(左旋细胞) 数量: $N_{\mathrm{LHR}}=N_{\mathrm{HR}}/2 = 30$
        - right HR cells(右旋细胞) 数量: $N_{\mathrm{RHR}}=N_{\mathrm{HR}}/2 = 30$

- 放电率
    - $t$ 时刻 HD firing rate(头朝向细胞放电率) $\vec{r}_{\mathrm{HD}}(t) \in \mathbb{R}^{N_{\mathrm{HD}}}$
        - 第 $i$ 个 HD cell 的 firing rate: $r_{\mathrm{HD},i}(t)$
    - $t$ 时刻 HR firing rate(头旋转细胞放电率) $\vec{r}_{\mathrm{HR}}(t) \in \mathbb{R}^{N_{\mathrm{HR}}}$
        - 第 $j$ 个 HR cell 的 firing rate: $r_{\mathrm{HR},j}(t)$

- 权重连接矩阵
    - HD $\to$ HD: **rec**urrent matrix, $\mathbf{W}_{\mathrm{HD}\to\mathrm{HD}}\in\mathbb{R}^{N_{\mathrm{HD}}\times N_{\mathrm{HD}}}$, 可塑
    - HR $\to$ HD: $\mathbf{W}_{\mathrm{HR}\to\mathrm{HD}} \in \mathbb{R}^{N_{\mathrm{HD}}\times N_{\mathrm{HR}}}$, 可塑
    - HD $\to$ HR: $\mathbf{W}_{\mathrm{HD}\to\mathrm{HR}} \in \mathbb{R}^{N_{\mathrm{HR}}\times N_{\mathrm{HD}}}$, 处理为 fixed

真实头朝向 $\theta(t)\in [0,2\pi]$, 真实角速度 $v(t) \in \mathbb{R}$

第 $i$ 个 HD cell 的 preferred direction(偏好方向) $\theta_{\mathrm{HD},i} \in [0,2\pi]$, 均匀分布在 ring 上. 

- Visual input(视觉输入) 矢量 $\vec{I}_{\mathrm{vis}\to\mathrm{HD}} \in \mathbb{R}^{N_{\mathrm{HD}}}$: periodic Gaussian-like bump
    - 视觉输入被传递到 HD cell 的 axon-proximal compartment(轴突近端区)
    - 第 $i$ 个分量是围绕真实头朝向 $\theta(t)$ 的 local bump: $\begin{aligned}I_{\mathrm{vis}\to\mathrm{HD},i}(t) = A_{\mathrm{vis}}\exp{\left[-\frac{1}{2\sigma_{\mathrm{vis}}^{2}}\sin^{2}\left(\frac{\theta_{\mathrm{HD},i}-\theta(t)}{2}\right)\right]}\end{aligned}$
        - $A_{\mathrm{vis}} \in \mathbb{R}$: 视觉 bump 幅值
        - $\sigma_{\mathrm{vis}} \in \mathbb{R}^{+}$: 视觉 bump 宽度
        - $I_{\mathrm{vis},0} \in \mathbb{R}^{-}$: 抑制性 baseline

- 角速度输入矢量 $\vec{I}_{\mathrm{vel}\to\mathrm{HR}}(t) \in \mathbb{R}^{N_{\mathrm{HR}}}$
    - (角)速度输入被传递到 HR cell
    - 第 $j$ 个分量为 $I_{\mathrm{vel}\to\mathrm{HR},j}(t) = q_{\mathrm{HR},j}k_{\mathrm{vel}}v(t)$
        - $k_{\mathrm{vel}}\in\mathbb{R}^{+}$: 速度到输入电流的比例系数
        - $\begin{aligned}q_{\mathrm{HR},j} = \begin{cases} -1, j\leq N_{\mathrm{HR}}/2 \\  +1, j > N_{\mathrm{HR}}/2\end{cases}\end{aligned}$, 用于区分 left HR 和 right HR. 











HD distal input current(HD 远端输入电流) 矢量 $\vec{I}_{\mathrm{HD},d}(t) \in \mathbb{R}^{N_{\mathrm{HD}}}$

- 动力学: $\begin{aligned}\tau_{s}\frac{\mathrm{d}\vec{I}_{\mathrm{HD},d}}{\mathrm{d}t} = -\vec{I}_{\mathrm{HD},d} + \mathbf{W}_{\mathrm{HD}\to\mathrm{HD}}\vec{r}_{\mathrm{HD}} + \mathbf{W}_{\mathrm{HR}\to\mathrm{HD}}\vec{r}_{\mathrm{HR}} + I_{\mathrm{inh},\mathrm{HD}}\vec{1}_{N_{\mathrm{HD}}}\end{aligned}$
    - $\tau_{s}$: synaptic time constant(突触时间常数)
    - $I_{\mathrm{inh},\mathrm{HD}}\in\mathbb{R}$: HD 集群的 global inhibition(全局抑制), $\vec{1}_{N_{\mathrm{HD}}}\in\mathbb{R}^{N_{\mathrm{HD}}} = \begin{bmatrix} 1 \\ 1 \\ \vdots \\ 1 \end{bmatrix}$








- HD cell 被处理为 2-compartment neuron(双室神经元)
    - $\tau_s\dot{\vec I}_{\mathrm{HD},d}=-\vec I_{\mathrm{HD},d}+\mathbf W_{\mathrm{HD}\to\mathrm{HD}}\vec r_{\mathrm{HD}}+\mathbf W_{\mathrm{HR}\to\mathrm{HD}}\vec r_{\mathrm{HR}}-b_{\mathrm{HD}}\vec 1$
    - HD axon-distal voltage(远端树突电压) 矢量 $\vec{V}_{\mathrm{HD},d}(t) \in \mathbb{R}^{N_{\mathrm{HD}}}$
        - $\tau_{l,\mathrm{HD}}\dot{\vec V}_{\mathrm{HD},d}=-\vec V_{\mathrm{HD},d}+\vec I_{\mathrm{HD},d}$；当前实现不令 $V_d=I_d$。
    - HD distal predition(远端预测) 矢量 $\vec{V}_{\mathrm{HD},ss}(t) = p_{d\to a}\vec{V}_{\mathrm{HD},d}(t)$
        - $\begin{aligned}p_{d\to a} = \frac{g_{d\to a}}{g_{d\to a} + g_{L}}\in(0,1)\end{aligned}$
            - $g_{d\to a}$: distal-to-proximal compling conductance(远端-近端耦合电导)
            - $g_{L}$: leak conductance(漏电导)
    - HD axon-proximal voltage(近端树突电压) 矢量 $\vec{V}_{\mathrm{HD},a}(t) \in \mathbb{R}^{N_{\mathrm{HD}}}$
        - 论文 Eq. (4)：$C\dot{\vec V}_{\mathrm{HD},a}=-g_L\vec V_{\mathrm{HD},a}-g_D(\vec V_{\mathrm{HD},a}-\vec V_{\mathrm{HD},d})+\vec I_{\mathrm{vis}\to\mathrm{HD}}+\vec I_{\mathrm{exc}}$。
    - sigmoid activation function(激活函数) $\begin{aligned}f(x) = \frac{f_{\max}}{1 + \exp{[-\beta(x-x_{1/2})]}}\end{aligned}$
        - $f_{\max}$: 最大 firing rate
        - $\beta$: 斜率
        - $x_{1/2}$: 半最大输入
    - HD firing rate(头朝向细胞放电率) 矢量 $\vec{r}_{\mathrm{HD}}(t) = f(\vec{V}_{\mathrm{HD},a}(t)) \in \mathbb{R}^{N_{\mathrm{HD}}}$

- HR cell 被处理为 single-compartment neuron(单室神经元)
    - 先低通 HD presynaptic rate：$\tau_s\dot{\vec r}_{\mathrm{HD}\to\mathrm{HR}}^{LP}=-\vec r_{\mathrm{HD}\to\mathrm{HR}}^{LP}+\vec r_{\mathrm{HD}}$。
    - HR input current 随后代数计算：$\vec I_{\mathrm{HR}}=\mathbf W_{\mathrm{HD}\to\mathrm{HR}}\vec r_{\mathrm{HD}\to\mathrm{HR}}^{LP}+\vec I_{\mathrm{vel}\to\mathrm{HR}}-b_{\mathrm{HR}}\vec 1$。
        - $I_{\mathrm{inh},\mathrm{HR}}$: HR 集群的 global inhibition
    - HR firing rate(头旋转细胞放电率) 矢量 $\vec{r}_{\mathrm{HR}}(t) = f(\vec{I}_{\mathrm{HR}}(t)) \in \mathbb{R}^{N_{\mathrm{HR}}}$




# Local learning rule

> 对于 plastic synapse $j\to i$, postsynaptic neuron $i$ 必须是 HD cell. ($\mathbf{W}_{\mathrm{HD}\to\mathrm{HD}}$ 和 $\mathbf{W}_{\mathrm{HR}\to\mathrm{HD}}$)

- postsynaptic learning error(突触后学习误差): $E_{\mathrm{HD},i}(t) = f(V_{\mathrm{HD},a,i}(t)) - f(V_{\mathrm{HD},ss,i}(t)) \in \mathbb{R}$
    - $f(V_{\mathrm{HD},a,i}(t))$: visual teacher 诱导的真实输出
    - $f(V_{\mathrm{HD},ss,i}(t))$: distal input 单独预测出的输出

- presynaptic postsynaptic potential(突触前诱发的突触后电位) $P_{\mathrm{pre},j}(t)\in \mathbb{R}$
    - $j\to i$ 由 presynaptic neuron $j$ 产生的局部 post-synaptic potential(PSP)
    - 动力学: $\begin{aligned}\tau_{P}\frac{\mathrm{d}P_{\mathrm{pre},j}}{\mathrm{d}t} = -P_{\mathrm{pre},j} + r_{\mathrm{pre},j}\end{aligned}$
        - $r_{\mathrm{pre},j}$: presynaptic neuron 的 firing rate

- learning rule: $\begin{aligned}\Delta W_{j\to i} = \eta E_{\mathrm{HD},i}(t)P_{\mathrm{pre},j} = \eta[f(V_{\mathrm{HD},a,i}(t)) - f(p_{d\to a}V_{\mathrm{HD},d,i}(t))]P_{\mathrm{pre},j}(t)\end{aligned}$
    
    原文形式为 $\Delta W_{pre, post} = \eta [f(V_{a,post}) - f(pV_{d,post})]P_{pre}$
    - $\eta>0$: learning rate(学习率)
    - $f(\vec{V}_{\mathrm{HD},a})$ 可通过 backpropagation of axonal activity(轴突活动的反向传播) 到达树突远端区

- recurrent weight update
    - HD PSP 矢量 $\vec{P}_{\mathrm{HD}}(t)\in\mathbb{R}^{N_{\mathrm{HD}}}$; Error 矢量 $\vec{E}_{\mathrm{HD}}(t)\in\mathbb{R}^{N_{\mathrm{HD}}}$
    - 动力学: $\begin{aligned}\frac{\mathrm{d}\mathbf{W}_{\mathrm{HD}\to\mathrm{HD}}}{\mathrm{d}t} = \eta_{\mathrm{HD}\to\mathrm{HD}} \vec{E}_{\mathrm{HD}}(t)[\vec{P}_{\mathrm{HD}}(t)]^{\top}\end{aligned}$
- HR-to-HD weight update
    - HR PSP 矢量 $\vec{P}_{\mathrm{HR}}(t)\in\mathbb{R}^{N_{\mathrm{HR}}}$
    - 动力学: $\begin{aligned}\frac{\mathrm{d}\mathbf{W}_{\mathrm{HR}\to\mathrm{HD}}}{\mathrm{d}t} = \eta_{\mathrm{HR}\to\mathrm{HD}} \vec{E}_{\mathrm{HD}}(t)[\vec{P}_{\mathrm{HR}}(t)]^{\top}\end{aligned}$









# Decode head direction

- population vector average(群体矢量平均, PVA): $\begin{aligned}z_{\mathrm{PVA}}(t) = \frac{1}{N_{\mathrm{HD}}}\sum_{i=1}^{N_{\mathrm{HD}}}r_{\mathrm{HD},i}(t)e^{i\theta_{\mathrm{HD},i}}\end{aligned}$
    - 内部估计头朝向: $\begin{aligned}\hat{\theta}_{\mathrm{HD}}(t) = \arg\,z_{\mathrm{PVA}}(t)\end{aligned}$, 和 Clark 的 COM 其实一致

# Toy model 设计

> 本节以下内容以当前 `src/learning` 实现（`models/vafidis_toy.py::step_vafidis_toy`）为准。论文给出的是连续时间耦合方程；release code 和当前代码都用一个全局 $\Delta t$ 在线推进，但在一个 timestep 内按固定顺序读写状态。

## 当前实现的群体与连接

- 默认 $N_{\mathrm{HD}}=N_{\mathrm{HR}}=N=60$。60 个 HD cell 按 odd/even（0-based 的 even/odd）配成 30 个相同 preferred direction，而不是 60 个独立角度 bin。
- $\vec r_{\mathrm{HD}}\in\mathbb R^N$；$\vec r_{\mathrm{HR}}=[\vec r_{\mathrm{LHR}};\vec r_{\mathrm{RHR}}]\in\mathbb R^N$，每个 HR wing 为 $\mathbb R^{N/2}$。
- $\mathbf W_{\mathrm{HD}\to\mathrm{HD}}\in\mathbb R^{N\times N}$；$\mathbf W_{\mathrm{HR}\to\mathrm{HD}}=[\mathbf W_{\mathrm{LHR}\to\mathrm{HD}},\mathbf W_{\mathrm{RHR}\to\mathrm{HD}}]\in\mathbb R^{N\times N}$，每个 wing 的权重块为 $N\times N/2$。
- 固定 $\mathbf W_{\mathrm{HD}\to\mathrm{HR}}\in\mathbb R^{N\times N}$ 是 sparse one-to-one selector：0-based even HD 投到第一半 HR，odd HD 投到第二半 HR；它不是两个 $N\times N$ identity 的纵向拼接。

## 连续时间含义

HD 的三个独立电学动力学状态是 distal current $\vec I_{\mathrm{HD},d}$、distal voltage $\vec V_{\mathrm{HD},d}$ 与 proximal voltage $\vec V_{\mathrm{HD},a}$：

$$\tau_s\dot{\vec I}_{\mathrm{HD},d}=-\vec I_{\mathrm{HD},d}+\mathbf W_{\mathrm{HD}\to\mathrm{HD}}\vec r_{\mathrm{HD}}+\mathbf W_{\mathrm{HR}\to\mathrm{HD}}\vec r_{\mathrm{HR}}-b_{\mathrm{HD}}\vec 1,$$

$$\tau_{l,\mathrm{HD}}\dot{\vec V}_{\mathrm{HD},d}=-\vec V_{\mathrm{HD},d}+\vec I_{\mathrm{HD},d}.$$

当前代码**没有**采用 $\vec V_{\mathrm{HD},d}=\vec I_{\mathrm{HD},d}$。TODO 中若把第二个低通量写作 $I_{\mathrm{HD}}$，它在论文/release code 与当前代码里的准确名称其实是 $V_{\mathrm{HD},d}$（代码字段 v_hd_distal）。近端视觉电流是 Eq. (4) 的输入，不是另一个独立积分的 current state。近端电压现在严格按

$$C\dot{\vec V}_{\mathrm{HD},a}=-g_L\vec V_{\mathrm{HD},a}-g_D(\vec V_{\mathrm{HD},a}-\vec V_{\mathrm{HD},d})+\vec I_{\mathrm{vis}\to\mathrm{HD}}+\vec I_{\mathrm{exc}}$$

积分，再计算 $\vec V_{\mathrm{HD},ss}=p_{d\to a}\vec V_{\mathrm{HD},d}$、$\vec r_{\mathrm{HD}}=f(\vec V_{\mathrm{HD},a})$ 与 $\vec E_{\mathrm{HD}}=f(\vec V_{\mathrm{HD},a})-f(\vec V_{\mathrm{HD},ss})$。这里只在初始化时把 $V_a$ 放到 Eq. (4) 的一致稳态；训练、关灯和测试都不会把它投影回 $V_{ss}$。Table 1 参数 $C=1\,\mathrm{ms},g_L=1,g_D=2$ 要求显式 Euler 满足 $\Delta t(g_L+g_D)/C<2$；baseline 使用 $\Delta t=0.5\,\mathrm{ms}$。

HR 通路只对 HD presynaptic rate 做一阶低通，速度输入即时相加：

$$\tau_{\mathrm{HD}\to\mathrm{HR}}\dot{\vec r}_{\mathrm{HD}\to\mathrm{HR}}^{LP}=-\vec r_{\mathrm{HD}\to\mathrm{HR}}^{LP}+\vec r_{\mathrm{HD}},$$

$$\vec I_{\mathrm{HR}}=\mathbf W_{\mathrm{HD}\to\mathrm{HR}}\vec r_{\mathrm{HD}\to\mathrm{HR}}^{LP}+\vec I_{\mathrm{vel}\to\mathrm{HR}}-b_{\mathrm{HR}}\vec 1,\qquad \vec r_{\mathrm{HR}}=f(\vec I_{\mathrm{HR}}).$$

代码的 wing 顺序是 $[\mathrm{LHR},\mathrm{RHR}]$，当前角度符号约定下 $\vec I_{\mathrm{vel}\to\mathrm{LHR}}=+k_vv\vec 1$、$\vec I_{\mathrm{vel}\to\mathrm{RHR}}=-k_vv\vec 1$。

每个 presynaptic population 的 PSP 是与论文 $H(t)$ 对应的两级滤波，而不是单一 $\tau_P$ 滤波：

$$\tau_s\dot{\vec P}_{\mathrm{syn}}=-\vec P_{\mathrm{syn}}+\vec r_{\mathrm{pre}},\qquad \tau_{l,\mathrm{HD}}\dot{\vec P}_{\mathrm{pre}}=-\vec P_{\mathrm{pre}}+\vec P_{\mathrm{syn}}.$$

两类权重使用同号的局部规则（HR 项没有额外负号）：

$$\tau_\delta\dot{\boldsymbol\delta}_{\mathrm{pre}\to\mathrm{HD}}=-\boldsymbol\delta_{\mathrm{pre}\to\mathrm{HD}}+\vec E_{\mathrm{HD}}\vec P_{\mathrm{pre}}^\top,\qquad \dot{\mathbf W}_{\mathrm{pre}\to\mathrm{HD}}=\eta_{\mathrm{pre}\to\mathrm{HD}}\boldsymbol\delta_{\mathrm{pre}\to\mathrm{HD}}.$$

当前 rule 没有 weight decay。Vafidis baseline 学习包括 diagonal 在内的完整 $\mathbf W_{\mathrm{HD}\to\mathrm{HD}}$；clipping、清零 diagonal、显式 symmetry/balance 都只在 config 明确启用时执行。

## 当前代码一个 timestep 的精确 data stream

令函数输入时的完整状态为 $S_n$，训练 driver 先产生本步使用的 $v_{n+1}$。`step_vafidis_toy` 按下列顺序构造 $S_{n+1}$：

1. 用 $v_{n+1}$ 更新 $\theta_{n+1}=\mathrm{wrap}(\theta_n+v_{n+1}\Delta t)$，并由 $\theta_{n+1}$ 产生 visual teacher、由 $v_{n+1}$ 产生 velocity input。
2. 先按 release network() 的顺序计算 HD distal pathway drive，明确读取旧 $\mathbf W^n$、旧 $\vec r_{\mathrm{HD}}^n$ 和旧 $\vec r_{\mathrm{HR}}^n$。
3. 用上述旧权重/旧 firing-rate drive 更新 $\vec I_{\mathrm{HD},d}^{n+1}$。
4. 用**刚更新的** $\vec I_{\mathrm{HD},d}^{n+1}$ Euler 更新 $\vec V_{\mathrm{HD},d}^{n+1}$；这是 timestep 内有序的 Gauss-Seidel-style 依赖，不是所有 RHS 都只读取 $S_n$ 的纯同步（Jacobi）Euler。
5. PSP 第一级读取旧 presynaptic rates $\vec r_{\mathrm{HD}}^n,\vec r_{\mathrm{HR}}^n$；第二级读取刚更新的 $\vec P_{\mathrm{syn}}^{n+1}$，得到 $\vec P_{\mathrm{HD}}^{n+1},\vec P_{\mathrm{HR}}^{n+1}$。
6. 用旧 $\vec r_{\mathrm{HD}}^n$ Euler 更新 $\vec r_{\mathrm{HD}\to\mathrm{HR}}^{LP,n+1}$；再用这个新 low-pass state 与本步 velocity input 代数计算 $\vec I_{\mathrm{HR}}^{n+1}$、$\vec r_{\mathrm{HR}}^{n+1}$。这个新 HR rate 不进入本步已经完成的 HD distal update。
7. 用旧 $\vec V_{\mathrm{HD},a}^{n}$、新 $\vec V_{\mathrm{HD},d}^{n+1}$ 和本步 proximal input，对 Eq. (4) 做一次 Euler 更新得到 $\vec V_{\mathrm{HD},a}^{n+1}$；同时计算 $\vec V_{\mathrm{HD},ss}^{n+1}=p\vec V_{\mathrm{HD},d}^{n+1}$、$\vec r_{\mathrm{HD}}^{n+1}$ 与 $\vec E_{\mathrm{HD}}^{n+1}$。
8. 训练时，用新 error 与新 PSP 得到 $\boldsymbol\delta^{n+1}$，再由旧 $\mathbf W^n$ 更新两类 $\mathbf W^{n+1}$，最后施加配置的 constraints。测试时权重和 $\boldsymbol\delta$ 冻结。
9. 将以上量一起封装为 $S_{n+1}$；新权重 $\mathbf W^{n+1}$ 到**下一个** step 才参与 $\vec I_{\mathrm{HD},d}$ 的 drive。

所以，连续时间层面是神经动力学与可塑性共同演化；数值实现层面是**同一时钟、每步一次更新、步内有序依赖**。它不是 event-driven asynchronous update，也没有“先固定权重等电流弛豫到稳态，再开始更新权重”的 alternating protocol。训练从第一步起每个 $\Delta t$ 都更新权重；$\tau_\delta$ 只负责把 plasticity induction 低通成较慢的学习过程。可选 early stopping 只决定何时结束整段训练，不是单步内的弛豫门控。

更精确地说，这是 single-clock ordered Euler：$I_{\mathrm{HD},d}^{n+1}\to V_{\mathrm{HD},d}^{n+1}\to V_{\mathrm{HD},a}^{n+1}\to E^{n+1}\to\Delta^{n+1}\to W^{n+1}$ 是步内新值依赖，而 distal drive 读取旧 $r_{\mathrm{HD}}^n,r_{\mathrm{HR}}^n,W^n$。权重无需等待电流或电位弛豫；$W^{n+1}$ 从下一步才影响 distal current。

冻结权重与视觉关闭时，Ramesan/slow-manifold 分析的最小 Markov 状态为 $x=[\vec r_{\mathrm{HD}\to\mathrm{HR}}^{LP},\vec r_{\mathrm{HR}},\vec I_{\mathrm{HD},d},\vec V_{\mathrm{HD},d},\vec V_{\mathrm{HD},a}]$；PVA 则仍只解码 $r_{\mathrm{HD}}=f(V_a)$。

仍保留的工程差异是峰值归一化 firing rate、秒单位与可复现 RNG；此外 HD→HR 低通遵循论文的标准 $\Delta t/\tau_s$ Euler，而不复制 release 文件中 x += (f-x)/tau_s 字面漏写 dt 的歧义。

## check

1. 给定短暂 visual pulse 后令 $v(t)=0$，检查解码位置 $\hat{\theta}_{\mathrm{HD}}$ 相比 $\theta_0$ 是否稳定。
2. 在冻结权重并移除持续 visual teacher 后，检查 $\hat{\theta}_{\mathrm{HD}}(t)\approx\theta(0)+\int_0^t v(t')\,\mathrm dt'$。
3. 用不同常数 $v$ 拟合 $\dot{\hat{\theta}}_{\mathrm{HD}}\approx gv$，理想结果为 $g\approx1$。
