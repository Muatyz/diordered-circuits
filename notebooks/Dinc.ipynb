# 0. 总纲

- ghost 动力学: $\dot{\kappa} = r + \kappa^{2}$
    
    $r>0$: 若 $r\to 0$, 则系统即使没有 fixed point, 但在 $\kappa = 0$ 附近运动极慢
    
    穿过 $\kappa\in [0, L]$ 所需时间 $\begin{aligned}t = \int_{0}^{L}\frac{\mathrm{d}\kappa}{\dot{\kappa}} = \int_{0}^{L}\frac{\mathrm{d}\kappa}{r + \kappa^{2}} = \frac{1}{\sqrt{r}}\arctan{\frac{L}{\sqrt{r}}}\end{aligned}$; 写作 $\kappa(t)$ 形式: $\kappa(t) = \sqrt{r}\tan{(\sqrt{r}t)}$
    
    极限形式 $\begin{aligned}t_{*} = \lim_{L\gg\sqrt{r}}\frac{T}{2}\approx \frac{\pi}{2\sqrt{r}}\end{aligned}$

    若网络需要等待时间 $T$, 需要 $t_{*}\sim T\Rightarrow r^{*}\sim T^{-2}$. $T$ 越长, $r^{*}$ 越接近 bifurcation point $r=0$

    critical learning rate $\alpha^{*}\sim T^{-4}$. 超过这个学习率, 则网络会进入 no-learning zone(gradient 几乎为 0) 

# 1. Intro

- Abrupt learning: 很长时间 loss 几乎不变, 随后突然学习成功
- 论点: 如果 temporal computation 是通过 ghost mechanism 实现的, 那么学习这个 mechanism 应当产生一般性的优化学

    > temporal computation 的实现并不局限于 ghost mechanism, 还有 fixed points, limit cycles... (特别是对 full-rank RNN)

# 2. Section II: Ghost Mechanism

## 2.1 slow point

- 状态变量: $\vec{x}\in\mathbb{R}^{N}$
- 动力学: $\dot{\vec{x}} = \vec{f}[\vec{x}(t)]$
- "动能"(state-space speed 的平方): $\begin{aligned}s(\vec{x}) = \frac{1}{2}||\vec{f}(\vec{x})||_{2}^{2}\end{aligned}$
- slow point: 对于 $\vec{x}^{*}$, 若满足
    - 速度局域不变: $\vec{\nabla} s(\vec{x}^{*}) = \vec{0}$

        > 仅有这个条件时, $\vec{x}^{*}$ 可能对 $s(\vec{x})$ 的极小值, 极大值或者 saddle point
    - 速度局域极小: $\nabla^{2} s(\vec{x}^{*}) \succ 0$

        > 表示 Hessian 正定(positive definite). $\forall\delta\vec{x}$, 都有 $\delta\vec{x}^{\top}\nabla^{2}s(\vec{x}^{*})\delta\vec{x} > 0$. 因此 Taylor 展开 $\begin{aligned}s(\vec{x}^{*} + \delta\vec{x}) \approx s(\vec{x}^{*}) + [\vec{\nabla} s(\vec{x})]^{\top}\delta\vec{x} + \frac{1}{2}\delta\vec{x}^{\top}\nabla^{2}s(\vec{x}^{*})\delta\vec{x} > s(\vec{x}^{*})\end{aligned}$
    - 速度最小但不为 0: $s(\vec{x}^{*}) > 0$

## 2.2 slow point 对应 zero mode

- "动能" 写作内积形式 $\begin{aligned}s(\vec{x}) = \frac{1}{2}[\vec{f}(\vec{x})]^{\top}\vec{f}(\vec{x})\end{aligned}$

    "动能" 梯度 $\vec{\nabla} s(\vec{x}) = [\mathbf{J}(\vec{x})]^{\top}\vec{f}(\vec{x})$

    > Jacobian 矩阵 $\begin{aligned}\mathbf{J}(\vec{x}) = \frac{\partial \vec{f}}{\partial\vec{x}}\end{aligned}$, 分量形式 $\begin{aligned}J_{ij} = \frac{\partial f_{i}}{\partial x_{j}}\end{aligned}$

    在 slow point $\vec{x}^{*}$, 有 $\vec{\nabla} s(\vec{x}^{*}) = [\mathbf{J}(\vec{x}^{*})]^{\top}\vec{f}(\vec{x}^{*}) = 0$
    
    令速度方向矢量 $\begin{aligned}\vec{v} = \frac{\vec{f}(\vec{x}^{*})}{||\vec{f}(\vec{x}^{*})||}\end{aligned}$, 有 $(\vec{v})^{\top}\mathbf{J}(\vec{x}^{*}) = 0$
    
    >  flow direction ($\vec{v}$) 对应 Jacobian 的一个 left zero mode

# 3. slow point $\neq$ ghost: transverse stability

- ghost: 垂直于 flow direction $\vec{v}$ 的方向 $\vec{w}\perp\vec{v}$, 若

    本征方程 $(\vec{w})^{\top}\mathbf{J} = \lambda (\vec{w})^{\top}$, 本征值满足 $\mathrm{Re}(\lambda) <0$, 则 slow point $\vec{x}^{*}$ 为 ghost

    > Ramesan 的 slow point set 未要求 transverse stability; Dinc's ghost 要求附近轨迹都被吸引到慢通道

# 4. Section II-B: High-dim ghost 约化为 1D

设 slow point 邻域 $\vec{z}(t) = \vec{x}(t) - \vec{x}^{*}$, 分解为

$$
\vec{z}(t) = \kappa(t)\vec{v} + \vec{z}_{\perp}(t)
$$

> $\vec{v}$: flow direction; $\vec{z}_{\perp}$: transverse direction. 两者满足 $(\vec{v})^{\top}\vec{z}_{\perp} = 0$
> 
> 通过 $\kappa = (\vec{v})^{\top}\vec{z}$ 来提取 "沿着 flow direction $\vec{v}$ 行走的分量"

- 推导
    - 时间导数: 
        $$\begin{aligned}\dot{\vec{z}} = \frac{\mathrm{d}}{\mathrm{d}t}(\vec{x}-\vec{x}^{*}) = \dot{\vec{x}} = \vec{f}(\vec{x}^{*} + \vec{z})\end{aligned}$$
    - $\vec{f}$ 在 $\vec{x}^{*}$ 处 Taylor 展开
        $$\vec{f}(\vec{x}^{*} + \vec{z}) = \vec{f}(\vec{x}^{*}) + \mathbf{J}(\vec{x}^{*})\vec{z} + \mathcal{O}(||\vec{z}||^{2})$$
    - 忽略二阶项
        $$\dot{z}\simeq \vec{f}(\vec{x}^{*}) + \mathbf{J}(\vec{x}^{*})\vec{z}$$

由于 transverse stability $\vec{z}_{\perp}(t)\to\vec{0}$, 则 high-dim state 被压缩为 $\vec{x}(t) \simeq \vec{x}^{*} + \kappa(t)\vec{v}$
        
沿 $\vec{v}$ 的动力学: $\dot{\kappa} = (\vec{v})^{\top}\dot{\vec{z}} = (\vec{v})^{\top}\vec{f}(\vec{x}^{*} + \vec{z})$

$\vec{f}$ 在 $\vec{x}^{*}$ 处 Taylor 展开: 

$$
\vec{f}(\vec{x}^{*} + \vec{z}) = 
\overset{\mathrm{0th}}{\vec{f}(\vec{x}^{*})} + 
\overset{\mathrm{1st}}{\mathbf{J}(\vec{x}^{*})(\vec{z})} + 
\overset{\mathrm{2nd}}{\frac{1}{2}\vec{h}[\vec{z},\vec{z}]} + 
\mathcal{O}(||\vec{z}||^{3})
$$

> $\vec{f}$ 有 $N$ 个分量, 各 $f_{i}(\vec{x})$ 都有自己的 Hessian $\begin{aligned}\mathbf{H}_{i} = \nabla^{2} f_{i} = \frac{\partial^{2} f_{i}}{\partial\vec{x}\partial\vec{x}}\end{aligned}$
> 
> 比如 $N = 2$ 时 $\begin{aligned}\vec{f} = \begin{bmatrix} f_1(x_{1}, x_{2}) \\ f_2(x_{1}, x_{2}) \end{bmatrix}\end{aligned}$, 两个 Hessian 为 $\begin{aligned}\mathbf{H}_{1} = \begin{bmatrix} \partial_{11}f_{1} & \partial_{12}f_{1} \\ \partial_{21}f_{1} & \partial_{22}f_{1} \end{bmatrix}\end{aligned}$ 和 $\begin{aligned}\mathbf{H}_{2} = \begin{bmatrix} \partial_{11}f_{2} & \partial_{12}f_{2} \\ \partial_{21}f_{2} & \partial_{22}f_{2} \end{bmatrix}\end{aligned}$, Taylor 二阶为 $\begin{aligned}\frac{1}{2}\begin{bmatrix}(\vec{z})^{\top}\mathbf{H}_{1}\vec{z} \\ (\vec{z})^{\top}\mathbf{H}_{2}\vec{z} \end{bmatrix}\end{aligned}$
> 
> 作者约定 $\begin{aligned}\begin{bmatrix} (\vec{z})^{\top}\mathbf{H}_{1}\vec{z} \\ \vdots \\ (\vec{z})^{\top}\mathbf{H}_{N}\vec{z} \end{bmatrix} = \vec{h}[\vec{z},\vec{z}]\end{aligned}$
 
代入 Taylor 展开至 $\dot{\kappa}$ 动力学: 

$$
\begin{aligned}
\dot{\kappa} &= (\vec{v})^{\top}\left[\vec{f}(\vec{x}^{*}) + \mathbf{J}(\vec{x}^{*})\vec{z} + \frac{1}{2}\vec{h}[\vec{z}, \vec{z}] + \mathcal{O}(||\vec{z}||^{3})\right]\\
&= (\vec{v})^{\top}\vec{f}(\vec{x}^{*}) + (\vec{v})^{\top}\mathbf{J}(\vec{x}^{*})\vec{z} + \frac{1}{2}(\vec{v})^{\top}\vec{h}[\vec{z}, \vec{z}] + \mathcal{O}(||\vec{z}||^{3})
\end{aligned}
$$

> 1. $\begin{aligned}(\vec{v})^{\top}\vec{f}(\vec{x}^{*}) = \frac{[\vec{f}(\vec{x}^{*})]^{\top}}{||\vec{f}(\vec{x}^{*})||}\vec{f}(\vec{x}^{*}) = \frac{||\vec{f}(\vec{x}^{*})||^{2}}{||\vec{f}(\vec{x}^{*})||} = ||\vec{f}(\vec{x}^{*})|| = \gamma > 0\end{aligned}$
> 
> 2. 由于 $\vec{\nabla}s = \mathbf{J}^{\top}\vec{f} = \vec{0}$, 所以 $\begin{aligned}\vec{v}^{\top}\mathbf{J} = \left(\mathbf{J}^{\top}\frac{\vec{f}}{||\vec{f}||}\right)^{\top} = (\vec{0})^{\top}\end{aligned}$, 因此第二项 $\vec{v}^{\top}\mathbf{J}\vec{z} = \vec{0}^{\top}\vec{z} = 0$

因此动力学没有一阶项: $\begin{aligned}\dot{\kappa} = \gamma + \frac{1}{2}\vec{v}^{\top}\vec{h}[\vec{z},\vec{z}]\end{aligned}$

- transverse direction $\vec{z}_{\perp} = \vec{z} - \kappa\vec{v}$; 动力学 $\dot{\vec{z}}_{\perp} = \dot{\vec{z}} - \dot{\kappa}\vec{v}$
    - 第一项: $\dot{\vec{z}}\simeq \vec{f}(\vec{x}^{*}) + \mathbf{J}\vec{z} = \vec{f}(\vec{x}^{*}) + \mathbf{J}(\kappa\vec{v} + \vec{z}_{\perp}) = \vec{f}(\vec{x}^{*}) + \kappa\mathbf{J}\vec{v} + \mathbf{J}\vec{z}_{\perp}$
    - 第二项: $\dot{\kappa}\vec{v}\simeq ||\vec{f}(\vec{x}^{*})||\vec{v} = \vec{f}(\vec{x}^{*})$ (忽略二阶项)
    - 合并: $\dot{\vec{z}}_{\perp} = \mathbf{J}\vec{z}_{\perp} + \kappa \mathbf{J}\vec{v}$
    - 因此 transverse component 会快速衰减到 $\vec{z}_{\perp}\simeq \kappa\vec{a}$

    > 证明: 

于是 $\vec{z} = \kappa\vec{v} + \kappa\vec{a} = \kappa(\vec{v} + \vec{a}) = \kappa\tilde{\vec{v}}$, 其中约定 $\tilde{\vec{v}} \equiv \vec{v} + \vec{a}$

Hessian quadratic form: $\vec{z}^{\top}\mathbf{H}_{i}\vec{z} = (\kappa\tilde{\vec{v}})^{\top}\mathbf{H}_{i}(\kappa\tilde{\vec{v}}) = \kappa^{2}\tilde{\vec{v}}^{\top}\mathbf{H}_{i}\tilde{\vec{v}}$

动力学: $\begin{aligned}\dot{\kappa} = \gamma + \frac{\kappa^{2}}{2}\vec{v}^{\top}\vec{h}[\tilde{\vec{v}}, \tilde{\vec{v}}] + \mathcal{O}(\kappa^3) = \gamma + \beta\kappa^{2} + \mathcal{O}(\kappa^3)\end{aligned}$

Rescale: $q = \beta\kappa$, 则 $\dot{q} = \beta\dot{\kappa}$, 代入有 $\dot{q} = \beta(\gamma + \beta\kappa^{2}) = \beta\gamma + q^{2}$, 定义 $r\equiv \beta\gamma$, 即变为 canonical form $\dot{q} = r + q^{2}$

# 5. Fig.1: Canonical ghost

<img src="https://files.seeusercontent.com/2026/08/20/u1Fy/20260820160808358.png" width="70%">

动力学 $\dot{\kappa} = r + \kappa^{2}$

- a. 
    - $r<0$ 对应 $\kappa_{\pm} = \pm\sqrt{-r}$ 两个 fixed points
    - $r = 0$ 时产生 saddle-node bifurcation
    - $r>0$ 产生 ghost. 设初态 $\kappa(0) =0$, 则 $\kappa(t) = \sqrt{r}\tan{(\sqrt{r}t)}$, 逃逸时间 $\begin{aligned}t_{*} = \frac{\pi}{2\sqrt{r}}\end{aligned}$
- b. 
    - 任务: $t\in[0,T]$ 输出 $0$, $t\in[T,2T]$ 输出 $1$. 写作 $o(t) = \begin{cases} 0, & t \in [0,T] \\ 1, & t \in [T,2T] \end{cases}$
    - 方法: 设定阈值 $\bar{\kappa}$, 利用阶跃函数 $\sigma(x) = \begin{cases} 1, & x \geq 0 \\ 0, & x < 0 \end{cases}$, 定义输出 $\hat{o}(\kappa) = \sigma[c(\kappa - \bar{\kappa})]$

        > 即在 ghost 区域附近($[0,\bar{\kappa}]$) 待够 $T$ 时间, 然后逃逸

# 6. Section II-D: 将动力学要求转为 loss 形式

- Loss function $\begin{aligned}L(r) = \int_{0}^{2T}[\hat{o}(\kappa(t)) - o(t)]^{2}\mathrm{d}t\end{aligned}$

- 研究 $c\to\infty, \bar{\kappa}\to\infty$ 的解析解
    - $\hat{o}(t) = \begin{cases} 0, t\in [0,t_{*}) \\ 1, t\in [t_{*},\infty) \end{cases}$
        - 逃逸时间 $\begin{aligned}t_{*} = \frac{\pi}{2\sqrt{r}}\end{aligned}$
    - 最优化条件即 $\begin{aligned}t_{*} = T\Rightarrow r^{*} = \frac{\pi^{2}}{4T^{2}}\propto T^{-2}\end{aligned}$
        - 长 $T$ 逼近 $r=0$. 



# 7. Fig.2: Optimization

delayed-activation(DA) task, canonical model 表现出 abrupt learning 和 learning stability

- 论点: 
    1. ghost mechanism 可以导致 abrupt learning
    2. abrupt learning 不要求训练过程中产生 bifurcation
    3. 同一 ghost loss landscape 自然产生 critical learning rate 和 no-learning zone

> canonical dynamics $\dot{\kappa} = r + \kappa^{2}$, 初态 $\kappa\bigg|_{t=0} = 0$. 逃逸时间 $\begin{aligned}t^{*}(r) = \frac{\pi}{2\sqrt{r}}\end{aligned}$. 输出 $\hat{o}(\kappa) = \sigma[c(\kappa-\bar{\kappa})]$
> 
> DA task: 目标输出 $o(t) = \begin{cases} 0, t\in [0,T) \\ 1, t\in [T,2T] \end{cases}$

参数设定 $T = 100,\, \Delta t = 0.1,\, \bar{\kappa} = 10$

<img src="https://files.seeusercontent.com/2026/08/21/hj0I/20260821164115551.png" width="40%">

- a. loss landscape
    - 理解 $r$ 对系统的影响
        - point of no return: 如果 $r$ 过小, 以至于整个 trial ($t\in [0,2T]$) 都没产生输出 $(\kappa(t)>\bar{\kappa})$, 则定义此时系统进入 no-learning zone($\nabla L = 0$). 对应解析解为 $\begin{aligned}t^{*} = \frac{\pi}{2\sqrt{r}} = 2T\Rightarrow r_{\mathrm{PNR}} = \frac{\pi^{2}}{16T^{2}} = \frac{r^{*}}{4}\end{aligned}$
        - global minimum: 即 $t^{*} = T$, 对应 $\begin{aligned}r^{*} = \frac{\pi^{2}}{4T^{2}}\end{aligned}$


    - Normalized loss values $\begin{aligned}L_{\mathrm{norm}}(r)\equiv \frac{L(r)}{T} = \frac{1}{T}\int_{0}^{2T}[\hat{o}(\kappa(t)) - o(t)]^{2}\mathrm{d}t\end{aligned}$

        > 此时 $L(r)$ 和 $L_{\mathrm{norm}}(r)$ 都是针对 $r$ 的定值函数, 还没有引入 learning. 
    - $L(r)$ 的解析解: 

        $$
        L(r) = \begin{cases}
        T, & \begin{aligned}r\in\left(0, \frac{r^{*}}{4}\right)\end{aligned} \\[2mm]
        \begin{aligned}\left|T-\frac{\pi}{2\sqrt{r}}\right|\end{aligned}, & \begin{aligned}r\in\left(\frac{r^{*}}{4},\infty\right)\end{aligned}
        \end{cases} \Rightarrow L_{\mathrm{norm}}(r) = \begin{cases}
        1, & \begin{aligned}r\in\left(0, \frac{r^{*}}{4}\right)\end{aligned} \\[2mm]
        \begin{aligned}\left|1-\frac{\pi}{2T\sqrt{r}}\right|\end{aligned}, & \begin{aligned}r\in\left(\frac{r^{*}}{4},\infty\right)\end{aligned}
        \end{cases}
        $$

        > $\nabla L(r)$ 的解析解: 
        > $$
        > \nabla L(r) = \begin{cases}
        > 0, & \begin{aligned}r\in\left(0, \frac{r^{*}}{4}\right)\end{aligned} \\[1mm]
        > \begin{aligned}-\frac{\pi}{4r^{3/2}}\end{aligned}, & \begin{aligned}r\in\left(\frac{r^{*}}{4}, r^{*}\right)\end{aligned} \\[1mm]
        > \begin{aligned}+\frac{\pi}{4r^{3/2}}\end{aligned}, & \begin{aligned}r\in\left(r^{*}, \infty\right)\end{aligned}
        > \end{cases}
        > $$

        注意到极限 $\begin{aligned}\lim_{r\gg r^{*}} L(r) = T\end{aligned}$ 和 $\begin{aligned}\lim_{r\gg r^{*}} \nabla L(r) = 0\end{aligned}$. 因此初始化 $r_{0}$ 过大会因为 $\nabla L$ 很小而学习缓慢, 维持高 $L$. 这也是 high-loss plateau 的成因




<img src="https://files.seeusercontent.com/2026/08/21/9Zjy/20260821164420475.png" width="80%">

- b.c. 初始化 $r = 10 r^{*}$, learning rate $\alpha = \{{\color{blue}{10^{-10}}}, {\color{orange}{10^{-9}}}\}$, gradient descent: $r_{n+1} = r_{n} - \alpha \nabla L(r_{n})$

    - $r > r^{*}$ 侧, $\nabla L > 0$, $r_{n+1} < r_{n}$; 而 $\begin{aligned}|\nabla L| = \frac{\pi}{4r^{3/2}}\end{aligned}$ 随着 $r\downarrow$ 而迅速 $\uparrow\uparrow$, 正反馈使得 $r$ 迅速下降. 这就是 abrupt learning 的成因. 

        > 同时注意到, 整个学习过程中可以在理论上始终维持 $r>0$, 因此并未产生 bifurcation
    - 注意 $r^{*}$ 是 $\nabla L$ 的间断点 $\begin{aligned}\left(\frac{\mathrm{d}L}{\mathrm{d}r}\bigg|_{r^{*-}} = -\frac{2T^{3}}{\pi^{2}}, \,\frac{\mathrm{d}L}{\mathrm{d}r}\bigg|_{r^{*+}} = \frac{2T^{3}}{\pi^{2}}\right)\end{aligned}$, 因此会因为 $- \alpha \nabla L(r_{n})$ 更新而跨过 $r^{*}$ 产生 oscillation. 

    - 若 $\alpha$ 过大, $- \alpha \nabla L(r_{n})$ 更新可能会使 $r$ 跨过 point of no return $\begin{aligned}\frac{r^{*}}{4}\end{aligned}$ 进入 no-learning zone
        - $\color{orange}{\alpha=10^{-9}}$ 过大, 在 $r^{*}$ 附近 oscillation 之后跨过 point of no return 而动弹不得
        - $\color{blue}{\alpha=10^{-10}}$ 较小, abrupt learning 到 $r^{*}$ 附近高频 oscillation

        > state-space dynamics 和 parameter-space dynamics 的连接

    - critical learning rate $\alpha^{*}$ 的导出
    
        要求一次 learning update 恰好跨过 optimum 右侧 到 point of no return 的距离: 
        
        $$
        \begin{aligned}
        \alpha^{*} |\nabla L(r^{*+})| = r^{*} - \frac{r^{*}}{4}\Rightarrow \alpha^{*}\frac{2T^{3}}{\pi^{2}} = \frac{3}{4}\frac{\pi^{2}}{4T^{2}}\Rightarrow \alpha^{*} = \frac{3\pi^{4}}{32}T^{-5}
        \end{aligned}
        $$

        根据模型参数可计算得到 $\alpha^{*} \approx 9.13\times 10^{-10}$. 这可以解释 ${\color{blue}{10^{-10}}} < \alpha^{*} < {\color{orange}{10^{-9}}}$ 两学习率分别成功和失败的原因

        > 对于 mean squared loss $\begin{aligned}L_{\mathrm{MSE}}\sim \frac{L}{T}\end{aligned}$, 对应 learning rate $\alpha_{\mathrm{MSE}}^{*}\propto T^{-4}$.

        无论是 $\alpha^{*}\propto T^{-5}$ 还是 $\alpha_{\mathrm{MSE}}^{*}\propto T^{-4}$, 都说明 $T\uparrow$, learning rate 需要 $\alpha\downarrow\downarrow$

<img src="https://files.seeusercontent.com/2026/08/21/Zfo1/20260821172916007.png" width="40%">

- d. learning rate 并不是越小越好. 
    - $\alpha\ll\alpha^{*}$: $-\alpha\nabla L$ 太小, 因此 $10r^{*}\to r^{*}$ 需要很长时间(更多 epochs). 
        - 低 epochs(1000) 下 loss 仍然很高 ($L/T\sim 1$)
        - 高 epochs(10000) 下 loss ($L/T$)才降低得很多
    - $\alpha\lesssim\alpha^{*}$: 最快的安全学习区域. 即使是低 epochs(1000) 也已经进入 $r^{*}$ 附近的 oscillation 区域
    - $\alpha\geq\alpha^{*}$: parameter($r$) 很容易进入 gradient = 0 的 no-learning zone, 因此 loss 居高不下($L/T = 1$)

# 8. Section III: 从 canonical model 到 low-rank RNN

> 从抽象的 canonical model $\dot{\kappa} = r + \kappa^{2}$ 跨越到具体的 RNN 模型

## 8.1 连续时间 firing-rate RNN: 

- 物理量
    - $N$: 神经元数量
    - $N_{\mathrm{in}}$: 输入神经元数量
    - $\vec{x}\in\mathbb{R}^{N}$: $N$ 个神经元的 firing-rate state
    - $\tau$: 神经时间常数
    - $\mathbf{W}\in\mathbb{R}^{N\times N}$: recurrent weight matrix
    - $\mathbf{W}_{\mathrm{in}}\in\mathbb{R}^{N\times N_{\mathrm{in}}}$: input weight matrix
    - $\vec{u}\in\mathbb{R}^{N_{\mathrm{in}}}$: 外部 input
    - $\vec{b}\in\mathbb{R}^{N}$: 神经元 bias
    - $\vec{\epsilon}\in\mathbb{R}^{N}$: neural noise(可设定为 $\vec{0}$)

- 动力学: $\boxed{\tau\dot{\vec{x}}(t) = -\vec{x}(t) + \tanh{\big[\mathbf{W}\vec{x}(t) + \mathbf{W}_{\mathrm{in}}\vec{u}(t) + \vec{b} + \vec{\epsilon}\big]}}$

> 这里 tanh 是 element-wise nonlinearity, 即 $\vec{z} = \begin{bmatrix} z_{1} \\ z_{2} \\ \vdots \\ z_{N} \end{bmatrix}$, 则 $\tanh{(\vec{z})} = \begin{bmatrix} \tanh(z_{1}) \\ \tanh(z_{2}) \\ \vdots \\ \tanh(z_{N}) \end{bmatrix}$

> 限制 $\mathbf{W}$ 为 rank-1 或 rank-2, 使得神经活动被约化为 low-dim latent dynamics

## 8.2 Section III A

### rank-1 RNN $\to$ 1D latent dynamics

rank-$K$ 的 $\mathbf{W}$ 具有性质 $\begin{aligned}\mathbf{W} = \sum_{k=1}^{K}\vec{m}^{(k)}(\vec{n}^{(k)})^{\top}\end{aligned}$

> [Example 1] 设 $\begin{aligned}\vec{m} = \begin{bmatrix} 2\\-1\\3\end{bmatrix}, \vec{n} = \begin{bmatrix} 1\\2\\-1\end{bmatrix}\end{aligned}$, 则构造 $\begin{aligned}\mathbf{W} = \vec{m}\vec{n}^{\top} = \begin{bmatrix} 2 & 4 & -2 \\ -1 & -2 & 1 \\ 3 & 6 & -3 \end{bmatrix} = [\vec{m}, 2\vec{m}, -1\vec{m}]\end{aligned}$
> 
> 即 $\mathbf{W}$ 的每一列都在同一方向 $\vec{m}$ 上, 因此 $\mathrm{rank}(\mathbf{W}) = 1$
> 
> 回到神经动力学, 设 firing-rate state $\begin{aligned}\vec{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix}\end{aligned}$, 则 $\begin{aligned}\mathbf{W}\vec{x} = (\vec{m}\vec{n}^{\top})\vec{x} = \vec{m}(\vec{n}^{\top}\vec{x}) = \vec{m} (n_{1}x_1 + n_{2}x_2 + n_{3}x_3)\end{aligned}$
> 
> 约定 $(n_{1}x_1 + n_{2}x_2 + n_{3}x_3) = \vec{n}^{\top}\vec{x} = \kappa$, 则 $\mathbf{W}\vec{x} = \vec{m} \kappa$, 这相当于通过 $\vec{n}^{\top}$ 将神经活动 $\vec{x}$ `read` 为 1D latent variable $\kappa$, 再通过 $\vec{m}$ `write` 到 $N$-dim recurrent input 中: 
> 
> $$
> \overset{N=3}{\vec{x}} \rightarrow \overset{K=1}{\vec{n}^{\top}\vec{x}} \rightarrow \overset{N=3}{\vec{m}\kappa}
> $$

> [Example 2] 同理, 构造 $\mathbf{W} = \vec{m}^{(1)}(\vec{n}^{(1)})^{\top} + \vec{m}^{(2)}(\vec{n}^{(2)})^{\top}$. 则 $\begin{aligned}\mathbf{W}\vec{x} = \vec{m}^{(1)} (\vec{n}^{(1)\top}\vec{x}) + \vec{m}^{(2)} (\vec{n}^{(2)\top}\vec{x}) = \vec{m}^{(1)}\kappa_{1} + \vec{m}^{(2)}\kappa_{2}\end{aligned}$. 
> 
> 更一般性地, recurrent feedback 可以约化为 $\begin{aligned}\mathbf{W}\vec{x} = \sum_{k=1}^{K}\vec{m}^{(k)}\vec{n}^{(k)\top}\vec{x} = \sum_{k=1}^{K}\vec{m}^{(k)}\kappa_{k}\end{aligned}$, 从而实现 data-stream: 
> 
> $$
> \vec{x} \overset{\vec{n}^{1\top}, \ldots, \vec{n}^{K\top}}{\longrightarrow} (\kappa_{1}, \kappa_{2}, \ldots, \kappa_{K}) \overset{\vec{m}^{1}, \ldots, \vec{m}^{K}}{\longrightarrow} \sum_{k=1}^{K}\vec{m}^{(k)}\kappa_{k}
> $$

> [Proof] rank-$K$ 矩阵进行 singular value decomposition(SVD): $\mathbf{W} = \mathbf{U}\mathbf{\Sigma}\mathbf{V}^{\top}$
> 
> 若 $\mathrm{rank}(\mathbf{W})=K\leq N$, 则存在 $K$ 个 non-zero singular values $\{\sigma_{1}, \cdots, \sigma_{K}\}$. 因此 SVD 也可写作 $\begin{aligned}\mathbf{W} = \sum_{k=1}^{K}\sigma_{k}\vec{u}_{k}\vec{v}_{k}^{\top} = [\vec{u}_{1}, \ldots, \vec{u}_{K}] \mathrm{diag}(\sigma_{1}, \ldots, \sigma_{K}) [\vec{v}_{1}, \ldots, \vec{v}_{K}]^{\top}\end{aligned}$ (这里采用了 compact SVD 写法)
> 
> 现构造 $\vec{m}^{(k)} = \sqrt{\sigma_{k}}\vec{u}_{k}$, $\vec{n}^{(k)} = \sqrt{\sigma_{k}}\vec{v}_{k}$, 即有 $\vec{m}^{(k)}(\vec{n}^{(k)})^{\top} = \sigma_{k}\vec{u}_{k}\vec{v}_{k}^{\top}$, 得证 $\begin{aligned}\mathbf{W} = \sum_{k=1}^{K}\vec{m}^{(k)}(\vec{n}^{(k)})^{\top}\end{aligned}$
> 
> 拆分方式并不唯一, 比如 $\vec{m}^{(k)} = \sigma_{k}\vec{u}_{k}$, $\vec{n}^{(k)} = \vec{v}_{k}$ 也可以使得原式成立

对于 $K=1$, 则 $\mathbf{W} = \vec{m}\vec{n}^{\top}$

定义 latent variable $\kappa(t) = \vec{n}^{\top}\vec{x}(t)$, 则 $\mathbf{W}\vec{x} = \vec{m}\kappa$

假定无输入无噪声($\vec{u}(t) = \vec{\epsilon} = \vec{0}$), 则动力学简化为 $\tau\dot{\vec{x}} = -\vec{x} + \tanh{(\mathbf{W}\vec{x} + \vec{b})}$. 

左乘 $\vec{n}^{\top}$, 得到面向 latent variable 的动力学: $\boxed{\tau\dot{\kappa}(t) = -\kappa(t) + \vec{n}^{\top}\tanh{[\vec{m}\kappa(t) + \vec{b}]}\equiv f(\kappa;\vec{m},\vec{n},\vec{b})}$

### latent RNN 和 canonical $r$ 的映射

动力学 $\tau\dot{\kappa}(t) = f(\kappa)$, 称作 latent flow. 假定训练后 $f(\kappa)$ 在 $\kappa = \kappa^{*}$ 处出现 local speed minimum $(f^{\prime}(\kappa^{*}) = 0, f^{\prime\prime}(\kappa^{*}) > 0)$

在 $\kappa = \kappa^{*}$ 附近 Taylor 展开: 

$$
\begin{aligned}
\tau\dot{\kappa}(t) = f(\kappa) &= {\color{blue}{f(\kappa^{*})}} + \cancel{f^{\prime}(\kappa^{*})(\kappa - \kappa^{*})} + {\color{red}{\frac{1}{2}f^{\prime\prime}(\kappa^{*})}}(\kappa - \kappa^{*})^{2} + \mathcal{O}((\kappa - \kappa^{*})^{3})\\
&= {\color{blue}{\gamma}} + {\color{red}{\beta}}(\kappa - \kappa^{*})^{2} + \mathcal{O}((\kappa - \kappa^{*})^{3})
\end{aligned}
$$

$f(\kappa) = -\kappa + \vec{n}^{\top}\tanh{[\vec{m}\kappa + \vec{b}]}$

> 设 $\vec{m} = \begin{bmatrix} m_{1} \\ m_{2} \\ \vdots \\ m_{N} \end{bmatrix}$, $\vec{n} = \begin{bmatrix} n_{1} \\ n_{2} \\ \vdots \\ n_{N} \end{bmatrix}$, $\vec{b} = \begin{bmatrix} b_{1} \\ b_{2} \\ \vdots \\ b_{N} \end{bmatrix}$. 则 $\tanh{(\vec{m}\kappa + \vec{b})} = \begin{bmatrix} \tanh{(m_{1}\kappa + b_{1})} \\ \tanh{(m_{2}\kappa + b_{2})} \\ \vdots \\ \tanh{(m_{N}\kappa + b_{N})} \end{bmatrix}$
> 
> 内积 $\begin{aligned}\vec{n}^{\top}\tanh{[\vec{m}\kappa + \vec{b}]} = \sum_{i=1}^{N}n_{i}\tanh{(m_{i}\kappa + b_{i})}\end{aligned}$, 于是 $f(\kappa)$ 写作普通一元函数 $\begin{aligned}f(\kappa) = -\kappa + \sum_{i=1}^{N}n_{i}\tanh{(m_{i}\kappa + b_{i})}\end{aligned}$

代入 $\kappa = \kappa^{*}$ 得到 $\begin{aligned}\boxed{\gamma = f(\kappa^{*}) = -\kappa^{*} + \sum_{i=1}^{N}n_{i}\tanh{(m_{i}\kappa^{*} + b_{i})}}\end{aligned}$

> 一阶导数 $f^{\prime}(\kappa)$: 
> 
> $$
> \begin{aligned}
> f^{\prime}(\kappa) &= -1 + \sum_{i=1}^{N}n_{i}\frac{\mathrm{d}}{\mathrm{d}\kappa} \tanh{(m_{i}\kappa + b_{i})} = -1 + \sum_{i=1}^{N}n_{i}\frac{\mathrm{d}\tanh[z_{i}(\kappa)]}{\mathrm{d}z_{i}}\frac{\mathrm{d}z_{i}(\kappa)}{\mathrm{d}\kappa} \\
> &= -1 + \sum_{i=1}^{N}n_{i}[1-\tanh^{2}(m_{i}\kappa + b_{i})] m_{i} = -1 + \sum_{i=1}^{N}n_{i}m_{i}[1-\tanh^{2}(m_{i}\kappa + b_{i})]
> \end{aligned}
> $$

$\kappa^{*}$ 满足 $f^{\prime}(\kappa^{*}) = 0$, 因此 $\begin{aligned}\sum_{i=1}^{N}n_{i}m_{i}[1-\tanh^{2}(m_{i}\kappa^{*} + b_{i})] = 1\end{aligned}$

> 二阶导数 $f^{\prime\prime}(\kappa)$: 
> 
> $$
> \begin{aligned}
> f^{\prime\prime}(\kappa) &= \sum_{i=1}^{N}n_{i}m_{i}\frac{\mathrm{d}}{\mathrm{d}\kappa}[1-\tanh^{2}(m_{i}\kappa + b_{i})] = \sum_{i=1}^{N}n_{i}m_{i}\frac{\mathrm{d}[1-\tanh^{2}(z_{i}(\kappa))]}{\mathrm{d}z_{i}}\frac{\mathrm{d}z_{i}(\kappa)}{\mathrm{d}\kappa} \\
> &=  \sum_{i=1}^{N}n_{i}m_{i}[-2\tanh(z_{i})(1-\tanh^{2}(z_{i}))] m_{i} \\
> &= -2\sum_{i=1}^{N}n_{i}m_{i}^{2}\tanh(m_{i}\kappa + b_{i})[1-\tanh^{2}(m_{i}\kappa + b_{i})]
> \end{aligned}
> $$

代入 $\kappa = \kappa^{*}$ 解得 $\begin{aligned}\frac{1}{2}f^{\prime\prime}(\kappa^{*})  = \boxed{\beta = -\sum_{i=1}^{N}n_{i}m_{i}^{2}\tanh(m_{i}\kappa^{*} + b_{i})[1-\tanh^{2}(m_{i}\kappa^{*} + b_{i})]}\end{aligned}$

类比 Section II 中的手法, 定义 canonical 参数 $r = \beta\gamma$. 这是一个很长的函数, 缩写为 $r = g(\vec{m}, \vec{n}, \vec{b};\kappa^{*}(\vec{m},\vec{n},\vec{b}))$


### RNN 的 gradient descent 如何变为 $r$ 的 gradient descent

rank-1 的 parameter space $\vec{\theta} = (\vec{m}, \vec{n}, \vec{b})$. canonical parameter 是参数空间为自变量的函数 $r = r(\vec{\theta})$

设参数更新 $\Delta\vec{\theta}$ 足够小. $r(\vec{\theta})$ 为 smooth function, 因此线性近似 $r(\vec{\theta} + \Delta\vec{\theta}) = r(\vec{\theta}) + \Delta r + \mathcal{O}(\|\Delta\vec{\theta}\|^2)$

> canonical update $\begin{aligned}\Delta r = \sum_{i=1}^{N}\frac{\partial r}{\partial m_{i}}\Delta m_{i} + \sum_{i=1}^{N}\frac{\partial r}{\partial n_{i}}\Delta n_{i} + \sum_{i=1}^{N}\frac{\partial r}{\partial b_{i}}\Delta b_{i} = [\vec{\nabla}_{\theta}\,r]^{\top}\Delta\vec{\theta}\end{aligned}$
> 
> 约定 $\vec{h}(\vec{\theta}) = \vec{\nabla}_{\theta} r$, 代表 $r$ 在 $\vec{\theta}$ 参数空间中变化最快的方向
> 
> 于是 $\Delta r = \vec{h}^{\top}\Delta\vec{\theta}$, 即 $\Delta r$ 是 $\Delta\vec{\theta}$ 在 $\vec{h} = \vec{\nabla}_{\theta} r$ 方向上的投影 [Example] $\Delta\vec{\theta}\perp\vec{h}$ 时, RNN 参数的确变了, 但是 canonical $r$ 却未被更新

- canonical model update: $\Delta r = -\alpha\nabla_{r} L(r)$
- rank-1 RNN updata: $\begin{aligned}\Delta m_{i} = -\alpha\frac{\partial L}{\partial m_{i}},\quad\Delta n_{i} = -\alpha\frac{\partial L}{\partial n_{i}},\quad\Delta b_{i} = -\alpha\frac{\partial L}{\partial b_{i}}\end{aligned}$. 联合得到 $\Delta\vec{\theta} = -\alpha\vec{\nabla}_{\theta}L$

联立得到关系 $\nabla_{r}L(r) = \vec{h}^{\top}\vec{\nabla}_{\theta}L(\vec{\theta})$

> 更严谨地使用 chain rule, 则 $L(\vec{\theta}) = L(r(\vec{\theta})) \Rightarrow \vec{\nabla}_{\theta}L = L^{\prime}(r) \vec{\nabla}_{\theta}r$, 同左乘 $\vec{h} = \vec{\nabla}_{\theta}r$ 有 $(\vec{\nabla}_{\theta}r)^{\top}\vec{\nabla}_{\theta}L = L^{\prime}(r)||\vec{\nabla}_{\theta}r||^{2}$

### Fig.4: 从 Delayed Cue Discrimination(DCD) 推广到 Working Memory

> DCD task: $\pm 1$ cue $\to T_{\mathrm{delay}}\to$ output cue

> Delayed Match-To-Sample(DMTS) task: $\mathrm{cue}_{1}\to T_{\mathrm{delay}}\to \mathrm{cue}_{2}\to \mathrm{match}/ \mathrm{nonmatch}$. 约定
> 
> - $T_{\mathrm{inp}}$: cue presentation duration(cue 展示期)
> - $T_{\mathrm{delay}}$: memory delay(延迟期)
> - $T_{\mathrm{resp}}$: response window(响应期)

> rank-1 训练不易收敛, 因此改用 rank-2 RNN. (这是一个 what+when 问题)

- latent variable $\kappa_{j}(t) = \vec{n}^{(j)\top}\vec{x}(t),\quad j=1,2$
- latent vector $\vec{\kappa} = \begin{bmatrix} \kappa_{1} \\ \kappa_{2} \end{bmatrix} = (\kappa_{1},\kappa_{2})^{\top}$ 
- latent dynamics

$$
\begin{aligned}
\tau\dot{\kappa}_{j} &= -\kappa_{j} + \vec{n}^{(j)\top}\tanh{[\vec{z}(t)]}\\
\text{effective drive}: \quad\vec{z}(t) &= \sum_{i=1}^{2}\vec{m}^{(i)}\kappa_{i}(t) + \mathbf{W}_{\mathrm{in}}\vec{u}(t) + \vec{b}\\
\text{output}:\quad \hat{o}(t) &= \psi[\vec{w}_{\mathrm{out}}^{\top}\vec{\kappa}(t) + \vec{b}_{\mathrm{out}}]
\end{aligned}
$$

> $\vec{w}_{\mathrm{out}}\in \mathbb{R}^{2}$: latent readout weights
> 
> $\vec{b}_{\mathrm{out}}\in \mathbb{R}^{2}$: latent readout bias
> 
> $\psi(\cdot)$: DCD 用 tanh, DMTS 用 sigmoid

第一次 abrupt learning: 形成 bistability(2-attractor), 学会 "what"

第二次 abrupt learning: 形成 ghosts, 学会 "when"
